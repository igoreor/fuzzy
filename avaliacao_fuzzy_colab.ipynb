# 🎤 Sistema de Avaliação de Apresentações com Lógica Fuzzy

## Projeto de Inteligência Artificial

### Objetivo
Desenvolver um sistema baseado em **Lógica Fuzzy** que permita avaliar apresentações orais a partir de parâmetros qualitativos, transformando avaliações subjetivas em graus quantitativos e gerando feedback construtivo.

### O que é Lógica Fuzzy?
A lógica fuzzy trabalha com **graus de verdade** entre 0 e 1, permitindo representar incertezas e informações imprecisas — algo mais próximo do raciocínio humano.

**Exemplo:** "A apresentação foi clara" pode ter valor 0.3, 0.7, 0.95, etc., dependendo da intensidade da clareza.

## 📚 Importação das Bibliotecas

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display, Markdown

print("✅ Bibliotecas importadas com sucesso!")
```

## 🔧 1. Definição das Variáveis

### Variáveis de Entrada (0-10)
1. **Clareza** - Quão clara foi a explicação do conteúdo
2. **Domínio** - Domínio do assunto pelo apresentador
3. **Ritmo** - Velocidade e fluidez da fala
4. **Materiais** - Qualidade e utilidade dos slides/recursos
5. **Engajamento** - Interação com a plateia, contato visual

### Variável de Saída (0-10)
- **Avaliação Final** - Nota final da apresentação

```python
# Criando as variáveis de entrada
clareza = ctrl.Antecedent(np.arange(0, 11, 0.1), 'clareza')
dominio = ctrl.Antecedent(np.arange(0, 11, 0.1), 'dominio')
ritmo = ctrl.Antecedent(np.arange(0, 11, 0.1), 'ritmo')
materiais = ctrl.Antecedent(np.arange(0, 11, 0.1), 'materiais')
engajamento = ctrl.Antecedent(np.arange(0, 11, 0.1), 'engajamento')

# Criando a variável de saída
avaliacao = ctrl.Consequent(np.arange(0, 11, 0.1), 'avaliacao')

print("✅ Variáveis criadas:")
print("   Entradas: Clareza, Domínio, Ritmo, Materiais, Engajamento")
print("   Saída: Avaliação Final")
```

## 📊 2. Variáveis Linguísticas e Funções de Pertinência

Cada variável possui **rótulos linguísticos** (ex.: Baixo, Médio, Alto) com **funções de pertinência** que definem o grau de verdade.

Usamos funções **triangulares** para simplicidade e interpretabilidade.

```python
# CLAREZA: Baixa, Média, Alta
clareza['baixa'] = fuzz.trimf(clareza.universe, [0, 0, 4])
clareza['media'] = fuzz.trimf(clareza.universe, [2, 5, 8])
clareza['alta'] = fuzz.trimf(clareza.universe, [6, 10, 10])

# DOMÍNIO: Fraco, Médio, Forte
dominio['fraco'] = fuzz.trimf(dominio.universe, [0, 0, 4])
dominio['medio'] = fuzz.trimf(dominio.universe, [2, 5, 8])
dominio['forte'] = fuzz.trimf(dominio.universe, [6, 10, 10])

# RITMO: Devagar, Adequado, Rápido
ritmo['devagar'] = fuzz.trimf(ritmo.universe, [0, 0, 4])
ritmo['adequado'] = fuzz.trimf(ritmo.universe, [3, 5, 7])
ritmo['rapido'] = fuzz.trimf(ritmo.universe, [6, 10, 10])

# MATERIAIS: Ruins, Aceitáveis, Bons
materiais['ruins'] = fuzz.trimf(materiais.universe, [0, 0, 4])
materiais['aceitaveis'] = fuzz.trimf(materiais.universe, [2, 5, 8])
materiais['bons'] = fuzz.trimf(materiais.universe, [6, 10, 10])

# ENGAJAMENTO: Baixo, Médio, Alto
engajamento['baixo'] = fuzz.trimf(engajamento.universe, [0, 0, 4])
engajamento['medio'] = fuzz.trimf(engajamento.universe, [2, 5, 8])
engajamento['alto'] = fuzz.trimf(engajamento.universe, [6, 10, 10])

# AVALIAÇÃO FINAL: 5 níveis
avaliacao['precisa_melhorar'] = fuzz.trimf(avaliacao.universe, [0, 0, 3])
avaliacao['aceitavel'] = fuzz.trimf(avaliacao.universe, [2, 4, 6])
avaliacao['bom'] = fuzz.trimf(avaliacao.universe, [5, 6.5, 8])
avaliacao['muito_bom'] = fuzz.trimf(avaliacao.universe, [7, 8.5, 10])
avaliacao['excelente'] = fuzz.trimf(avaliacao.universe, [8.5, 10, 10])

print("✅ Funções de pertinência definidas para todas as variáveis!")
```

### Visualização das Funções de Pertinência

```python
fig, axes = plt.subplots(3, 2, figsize=(15, 12))
fig.suptitle('Funções de Pertinência - Sistema de Avaliação de Apresentações', 
             fontsize=16, fontweight='bold')

clareza.view(ax=axes[0, 0])
axes[0, 0].set_title('Clareza')

dominio.view(ax=axes[0, 1])
axes[0, 1].set_title('Domínio do Conteúdo')

ritmo.view(ax=axes[1, 0])
axes[1, 0].set_title('Ritmo da Fala')

materiais.view(ax=axes[1, 1])
axes[1, 1].set_title('Qualidade dos Materiais')

engajamento.view(ax=axes[2, 0])
axes[2, 0].set_title('Engajamento')

avaliacao.view(ax=axes[2, 1])
axes[2, 1].set_title('Avaliação Final')

plt.tight_layout()
plt.show()
```

## 📝 3. Base de Regras Fuzzy

Definimos **15 regras** (mínimo exigido: 10) no formato:

**IF** [condições] **THEN** [consequência]

### Exemplos de Regras:
1. **IF** Clareza é Alta **AND** Domínio é Forte **AND** Engajamento é Alto **THEN** Avaliação é Excelente
2. **IF** Clareza é Baixa **AND** Domínio é Fraco **THEN** Avaliação é Precisa Melhorar
3. **IF** Ritmo é Devagar **OR** Ritmo é Rápido **AND** Engajamento é Baixo **THEN** Avaliação é Aceitável

```python
# Definindo as 15 regras
regra1 = ctrl.Rule(clareza['alta'] & dominio['forte'] & engajamento['alto'], 
                   avaliacao['excelente'])

regra2 = ctrl.Rule(clareza['alta'] & dominio['forte'] & materiais['bons'], 
                   avaliacao['muito_bom'])

regra3 = ctrl.Rule(clareza['media'] & dominio['medio'] & engajamento['medio'], 
                   avaliacao['bom'])

regra4 = ctrl.Rule(clareza['alta'] & dominio['forte'] & materiais['ruins'], 
                   avaliacao['bom'])

regra5 = ctrl.Rule((ritmo['devagar'] | ritmo['rapido']) & engajamento['baixo'], 
                   avaliacao['aceitavel'])

regra6 = ctrl.Rule(clareza['baixa'] & dominio['fraco'], 
                   avaliacao['precisa_melhorar'])

regra7 = ctrl.Rule(materiais['bons'] & clareza['media'] & dominio['medio'], 
                   avaliacao['bom'])

regra8 = ctrl.Rule(engajamento['alto'] & ritmo['adequado'], 
                   avaliacao['muito_bom'])

regra9 = ctrl.Rule(clareza['media'] & dominio['medio'] & materiais['aceitaveis'], 
                   avaliacao['aceitavel'])

regra10 = ctrl.Rule(clareza['baixa'] & dominio['fraco'] & engajamento['baixo'], 
                    avaliacao['precisa_melhorar'])

regra11 = ctrl.Rule(dominio['forte'] & clareza['media'] & ritmo['adequado'], 
                    avaliacao['bom'])

regra12 = ctrl.Rule(ritmo['adequado'] & clareza['alta'] & materiais['bons'], 
                    avaliacao['muito_bom'])

regra13 = ctrl.Rule(materiais['ruins'] & engajamento['baixo'] & clareza['baixa'], 
                    avaliacao['precisa_melhorar'])

regra14 = ctrl.Rule(clareza['media'] & ritmo['adequado'] & engajamento['medio'], 
                    avaliacao['aceitavel'])

regra15 = ctrl.Rule(engajamento['alto'] & dominio['forte'] & materiais['aceitaveis'], 
                    avaliacao['muito_bom'])

print("✅ 15 regras fuzzy criadas com sucesso!")
print("\nExemplos de regras:")
print("1. IF Clareza=Alta AND Domínio=Forte AND Engajamento=Alto THEN Excelente")
print("2. IF Clareza=Baixa AND Domínio=Fraco THEN Precisa Melhorar")
print("3. IF Engajamento=Alto AND Ritmo=Adequado THEN Muito Bom")
print("... e mais 12 regras!")
```

## ⚙️ 4. Criação do Sistema de Controle Fuzzy

Usamos o método de inferência **Mamdani** (clássico) com:
- **Agregação**: max (OR) e min (AND)
- **Defuzzificação**: Centro de gravidade (centroid)

```python
# Criando o sistema de controle
sistema_ctrl = ctrl.ControlSystem([
    regra1, regra2, regra3, regra4, regra5,
    regra6, regra7, regra8, regra9, regra10,
    regra11, regra12, regra13, regra14, regra15
])

# Criando a simulação
simulacao = ctrl.ControlSystemSimulation(sistema_ctrl)

print("✅ Sistema de controle fuzzy criado!")
print("   Método: Mamdani")
print("   Defuzzificação: Centroid (Centro de Gravidade)")
```

## 🎯 5. Função de Avaliação

```python
def avaliar_apresentacao(clareza_val, dominio_val, ritmo_val, materiais_val, engajamento_val, mostrar_detalhes=True):
    """
    Avalia uma apresentação usando lógica fuzzy.
    
    Parâmetros:
        clareza_val (float): 0-10
        dominio_val (float): 0-10
        ritmo_val (float): 0-10
        materiais_val (float): 0-10
        engajamento_val (float): 0-10
        mostrar_detalhes (bool): Se True, mostra análise detalhada
    
    Retorna:
        dict: Nota e classificação
    """
    # Define os valores de entrada
    simulacao.input['clareza'] = clareza_val
    simulacao.input['dominio'] = dominio_val
    simulacao.input['ritmo'] = ritmo_val
    simulacao.input['materiais'] = materiais_val
    simulacao.input['engajamento'] = engajamento_val
    
    # Executa a inferência fuzzy
    simulacao.compute()
    
    # Obtém o resultado
    nota = simulacao.output['avaliacao']
    
    # Classifica o resultado
    if nota < 3:
        classificacao = "Precisa Melhorar"
        emoji = "🔴"
    elif nota < 5:
        classificacao = "Aceitável"
        emoji = "🟡"
    elif nota < 7:
        classificacao = "Bom"
        emoji = "🟢"
    elif nota < 8.5:
        classificacao = "Muito Bom"
        emoji = "🔵"
    else:
        classificacao = "Excelente"
        emoji = "🟣"
    
    if mostrar_detalhes:
        print("=" * 70)
        print("📊 RESULTADO DA AVALIAÇÃO FUZZY")
        print("=" * 70)
        print(f"Clareza:      {clareza_val}/10")
        print(f"Domínio:      {dominio_val}/10")
        print(f"Ritmo:        {ritmo_val}/10")
        print(f"Materiais:    {materiais_val}/10")
        print(f"Engajamento:  {engajamento_val}/10")
        print("-" * 70)
        print(f"🎯 NOTA FINAL: {nota:.2f}/10")
        print(f"{emoji} CLASSIFICAÇÃO: {classificacao}")
        print("=" * 70)
    
    return {
        'nota': round(nota, 2),
        'classificacao': classificacao,
        'emoji': emoji
    }

print("✅ Função de avaliação criada!")
```
